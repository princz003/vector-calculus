# **Aim**

To compute the gradient of a given scalar field and visualize

the corresponding gradient vector field using Python.

# **Objective**

* To understand the concept of the gradient of a scalar field

* To compute the gradient using SymPy

* To evaluate the gradient numerically using NumPy

* To visualize the gradient vector field using Matplotlib

# THEORY:

The gradient of a scalar field f(x, y, z) is a vector field defined as:

              ∇f = (∂f/∂x, ∂f/∂y, ∂f/∂z)

 The gradient points in the direction of the maximum rate of increase

of the scalar field, and its magnitude represents the rate of change.

**CASE STUDY:**

Given the scalar field:

                f(x, y, z) = x^2 + y^2 + z^2

The gradient of the scalar field is:

∇f = (2x, 2y, 2z)

For visualization purposes, the gradient field is plotted
in the plane z = 0.



# **Algorithm**

* Step 1: Import the libraries Numpy, Sympy and Matplotlib

* Step 2: Define the symbols

* Step 3: Define the scalar field

* Step 4: Compute the gradient

* Step 5: Define a grid off points where you want to evaluate the gradient

* Step 6: Initialize arrays to store gradient components

* Step 7: Evaluate the gradient at each point

* Step 8: Plot the gradient vectors using matplotlib

> **Case** Find the gradient of the scalar field $f=x^2+y^2+z^2$ and plot the gradient field.

**Solution**

```python
# Import SymPy for symbolic mathematics
import sympy as sp

# Import NumPy for numerical computation and array handling
import numpy as np

# Import Matplotlib for plotting the vector field
import matplotlib.pyplot as plt


# STEP 1: Define symbolic variables x, y, and z
x, y, z = sp.symbols('x y z')


# STEP 2: Define the scalar field f(x, y, z) = x^2 + y^2 + z^2
scalar_field = x**2 + y**2 + z**2


# STEP 3: Compute the gradient of the scalar field
# Gradient = (∂f/∂x, ∂f/∂y, ∂f/∂z)
gradient = sp.Matrix([
    sp.diff(scalar_field, x),
    sp.diff(scalar_field, y),
    sp.diff(scalar_field, z)
])


# STEP 4: Create a grid of points for evaluation
# Generate evenly spaced values for x and y
# np.linspace(a, b, n) generates n evenly spaced points between a and b (inclusive)
x_vals = np.linspace(-10, 10, 20)
y_vals = np.linspace(-10, 10, 20)

# Fix z = 0 for 2D visualization
z_val = 0

# Create a mesh grid from x and y values
X, Y = np.meshgrid(x_vals, y_vals)


# STEP 5: Initialize arrays to store x and y components of gradient
U = np.zeros_like(X, dtype=float)
V = np.zeros_like(Y, dtype=float)


# STEP 6: Evaluate the gradient at each grid point
# Substitute numerical values of x, y, and z into the symbolic gradient
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        grad_val = gradient.subs({x: X[i, j], y: Y[i, j], z: z_val})
        U[i, j] = float(grad_val[0])  # x-component of gradient
        V[i, j] = float(grad_val[1])  # y-component of gradient
print("="*60)
print("="*60)
print("Sample Numerical Evaluation (z = 0):")
print("="*60)
print("Gradient at point (2, 3, 0):")
print("="*60)
sp.pprint(gradient.subs({x: 2, y: 3, z: 0}))
print("="*60)
print("="*60)


# STEP 7: Plot the gradient vector field using quiver plot
plt.figure()
plt.quiver(X, Y, U, V)

# Label the axes
plt.xlabel('X')
plt.ylabel('Y')

# Add title to the plot
plt.title('Gradient Vector Field of f = x² + y² + z² (z = 0)')

# Display grid for better visualization
plt.grid(True)

# Show the final plot
plt.show()
```

# **Application Problem:**

Temperature Distribution in a Metal Plate

Consider a thin metal plate with uniform thickness. The temperature distribution across

the plate is described by a scalar field T(x, y), where T is the temperature at the point

(x, y) on the plate. The temperature distribution is given by the function:

T(x, y) = 100 e^{-(x^2 + y^2)/2}

where the temperature is measured in degrees Celsius and the coordinates x and y are

measured in meters.

(i) Determine the temperature gradient ∇T(x, y), which describes the rate and direction

of the most rapid temperature increase at any point (x, y) on the plate.

(ii) Compute and visualize the temperature distribution T(x, y) over a specified region

of the plate.

(iii) Compute and visualize the temperature gradient ∇T(x, y) over the same region

of the plate.




```python
# ============================================================
# APPLICATION PROBLEM 1: TEMPERATURE DISTRIBUTION IN A METAL PLATE
# ============================================================

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

print("="*60)
print("APPLICATION PROBLEM 1: TEMPERATURE DISTRIBUTION")
print("="*60)

# Define symbolic variables
x, y = sp.symbols('x y')

# Define temperature distribution
T = 100 * sp.exp(-(x**2 + y**2)/2)

print("Temperature Distribution T(x, y):")
sp.pprint(T)

print("="*60)

# Compute temperature gradient
grad_T = sp.Matrix([sp.diff(T, x), sp.diff(T, y)])

print("Temperature Gradient ∇T:")
sp.pprint(grad_T)

print("="*60)

# Numerical grid
x_vals = np.linspace(-3, 3, 30)
y_vals = np.linspace(-3, 3, 30)
X, Y = np.meshgrid(x_vals, y_vals)

# Evaluate temperature
T_func = sp.lambdify((x, y), T, 'numpy')
T_vals = T_func(X, Y)

# Evaluate gradient
grad_Tx = sp.lambdify((x, y), grad_T[0], 'numpy')
grad_Ty = sp.lambdify((x, y), grad_T[1], 'numpy')

U = grad_Tx(X, Y)
V = grad_Ty(X, Y)

print("Sample Temperature at (0,0):", T.subs({x:0, y:0}))
print("Sample Gradient at (1,1):")
sp.pprint(grad_T.subs({x:1, y:1}))

print("="*60)

# Plot temperature distribution
plt.figure()
plt.contourf(X, Y, T_vals)
plt.colorbar(label='Temperature (°C)')
plt.title("Temperature Distribution T(x, y)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

# Plot temperature gradient
plt.figure()
plt.quiver(X, Y, U, V)
plt.title("Temperature Gradient Field ∇T")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

# **Application Problem:**

Electric Potential Distribution in a Region

Consider a two-dimensional region with a static electric field. The electric potential

V(x, y) in this region is given by a scalar field that describes the potential at any

point (x, y). The electric potential distribution is given by the function:

V(x, y) = V₀ e^{-(x^2 + y^2)/(2a^2)}

where V₀ is the maximum potential at the origin (0, 0) and a is a characteristic
length scale.

(i) Determine the electric field E(x, y), which is defined as the negative gradient

of the electric potential, E = −∇V.

(ii) Compute and visualize the electric potential distribution V(x, y) over a specified

region.

(iii) Compute and visualize the electric field E(x, y) over the same region.


```python

# ============================================================
# APPLICATION PROBLEM 2: ELECTRIC POTENTIAL DISTRIBUTION
# ============================================================

print("="*60)
print("APPLICATION PROBLEM 2: ELECTRIC POTENTIAL DISTRIBUTION")
print("="*60)

# Define constants
V0, a = sp.symbols('V0 a', positive=True)

# Define electric potential
V = V0 * sp.exp(-(x**2 + y**2)/(2*a**2))

print("Electric Potential V(x, y):")
sp.pprint(V)

print("="*60)

# Electric field E = -∇V
E = -sp.Matrix([sp.diff(V, x), sp.diff(V, y)])

print("Electric Field E = -∇V:")
sp.pprint(E)

print("="*60)

# Assign numerical values
V0_val = 10
a_val = 1

V_num = V.subs({V0: V0_val, a: a_val})
E_num = E.subs({V0: V0_val, a: a_val})

# Lambdify
V_func = sp.lambdify((x, y), V_num, 'numpy')
Ex_func = sp.lambdify((x, y), E_num[0], 'numpy')
Ey_func = sp.lambdify((x, y), E_num[1], 'numpy')

V_vals = V_func(X, Y)
Ex_vals = Ex_func(X, Y)
Ey_vals = Ey_func(X, Y)

print("Sample Electric Potential at (0,0):", V_num.subs({x:0, y:0}))
print("Sample Electric Field at (1,1):")
sp.pprint(E_num.subs({x:1, y:1}))

print("="*60)

# Plot electric potential
plt.figure()
plt.contourf(X, Y, V_vals)
plt.colorbar(label='Electric Potential (V)')
plt.title("Electric Potential Distribution V(x, y)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()

# Plot electric field
plt.figure()
plt.quiver(X, Y, Ex_vals, Ey_vals)
plt.title("Electric Field Distribution E(x, y)")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
